```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print(pd.__version__, np.__version__)
```

```python
data = {
    "Name": ["Kajal","Riya","Amit","Sneha","Rahul","Pooja","Neha","Karan"],
    "Age": [20,21,20,22,21,20,22,21],
    "Marks": [65,78,45,88,52,91,70,38],
    "City": ["Pune","Pune","Mumbai","Nashik","Pune","Mumbai","Nashik","Pune"]
}
df = pd.DataFrame(data)
df
```

```python
df.describe()  # mean, min, max
```

```python
df[df["Marks"] > 60]
```

```python
df.sort_values("Marks", ascending=False)   # sort
```

```python
df["Result"] = np.where(df["Marks"] >= 40, "Pass", "Fail")   # new column
```

```python
df.groupby("City")["Marks"].mean() # average marks per city
```

```python
df.isnull().sum()                  # check missing values
```

```python
plt.bar(df["Name"], df["Marks"]); plt.title("Marks of Students"); plt.show()
df["City"].value_counts().plot(kind="pie"); plt.show()
plt.hist(df["Marks"]); plt.title("Marks Distribution"); plt.show()
```

```python
print(df["Marks"].mean(), df["Marks"].median(), df["Marks"].std())
print(df["Marks"].max(), df["Marks"].min())
```
